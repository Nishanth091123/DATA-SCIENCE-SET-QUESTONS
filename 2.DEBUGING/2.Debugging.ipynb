## 3. Outlier Detection using IQR

```python
#Incorrect Code
import numpy as np

data = [10, 12, 14, 16, 18, 100]

# Error: Incorrect IQR calculation
Q1, Q3 = np.percentile(data, 0.25, 0.75)
IQR = Q3 - Q1
outliers = [x for x in data if x < (Q1 - 1.5 * IQR) or x > (Q3 + 1.5 * IQR)]

print("Outliers:", outliers)
```

```python
import numpy as np

data = [10, 12, 14, 16, 18, 100]

# Calculate Q1 and Q3
Q1, Q3 = np.percentile(data, [25, 75])

# Calculate IQR
IQR = Q3 - Q1

# Find outliers
outliers = [
    x for x in data
    if x < (Q1 - 1.5 * IQR) or x > (Q3 + 1.5 * IQR)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

```python
from scipy import stats

medication_group = [110, 115, 108, 102, 107]
placebo_group = [120, 125, 130, 122, 128]

# Independent t-test
t_stat, p_value = stats.ttest_ind(
    medication_group,
    placebo_group
)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Sample data
data = pd.DataFrame({
    'Age': [5, 3, 8, 2, 1, 7, 4, 6, 3, 9],
    'Mileage': [50000, 30000, 80000, 20000, 15000, 70000, 40000, 60000, 35000, 90000],
    'EngineSize': [1.2, 1.6, 2.0, 1.0, 1.3, 2.5, 1.8, 2.2, 1.5, 2.8]
})

# Calculate VIF
vif_data = pd.DataFrame()

vif_data['Feature'] = data.columns

vif_data['VIF'] = [
    variance_inflation_factor(data.values, i)
    for i in range(data.shape[1])
]

print(vif_data)
```
