## Debugging 1: Correlation Heatmap

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Experience': [1, 3, 5, 7, 10],
    'Projects': [2, 5, 8, 10, 12],
    'Performance': [3, 6, 7, 8, 9],
    'Department': ['HR', 'IT', 'Sales', 'IT', 'HR']
})
correlation_matrix = data.select_dtypes(include='number').corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## Debugging 2: Missing Values

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Name': ['John', 'Emily', 'Michael', 'kishor', 'Sarah'],
    'Age': [25, 30, np.nan, 28, 22],
    'Salary': [50000, 60000, 55000, 58000, np.nan]
})
for column in ['Age', 'Salary']:
    data[column] = data[column].fillna(data[column].mean())
print(data)
```

## Debugging 3: Independent-Samples T-Test

```python
from scipy import stats

group1 = [120, 125, 130, 122, 128]
group2 = [110, 115, 108, 102, 107]
t_stat, p_value = stats.ttest_ind(group1, group2)
print(f'T-Statistic: {t_stat:.2f}, P-Value: {p_value:.4f}')
if p_value <= 0.05:
    print('Reject the null hypothesis: Significant difference.')
else:
    print('Fail to reject the null hypothesis: No significant difference.')
```

## Debugging 4: Salary Outliers

```python
import pandas as pd

data = pd.DataFrame({'Salary': [30000, 32000, 34000, 500000,
    36000, 37000, 39000, 41000, 42000, 450000]})
q1 = data['Salary'].quantile(0.25)
q3 = data['Salary'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = data[(data['Salary'] < lower) | (data['Salary'] > upper)]
print('Outliers:')
print(outliers)
```

## Debugging 5: Histogram / PDF Plot

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
data = np.random.normal(50, 10, 1000)
plt.hist(data, bins=30, density=True, alpha=0.6, edgecolor='black')
sns.kdeplot(data)
plt.title('Probability Density Plot')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```
